```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data.csv")
```

```python
countries = df["Country"].unique()
```

```python
plt.figure()

for country in countries:
    df_country = df[df["Country"] == country]
    plt.plot(df_country["Year"], df_country["Alcohol"], marker="o", linestyle="-", label=country, alpha=1)

plt.xlabel("Year")
plt.ylabel("Alcohol consumption per country")
# plt.legend(loc="upper right", fontsize="small", bbox_to_anchor=(1.3, 1))
plt.grid(True)
plt.show()
```

```python
import seaborn as sns
```

```python
plt.figure()
sns.lineplot(data=df, x = "Year", y = "Alcohol", hue="Country", alpha=1)
plt.legend(loc="upper left", bbox_to_anchor=(1,1))
plt.xlabel("Year")
plt.ylabel("Alcohol consumption per country")
plt.show()
```

```python
import matplotlib.pyplot as plt
```

```python
years = [2015, 2016, 2017, 2018, 2019, 2020]
sales = [50, 55, 53, 60, 67, 80] # in thousands

plt.plot(years, sales, marker='o', linestyle='-.', color='b', label='Sales over Years')
plt.xlabel('Year')
plt.ylabel('Sales (in thousands)')
plt.title("Yearly sales growth")
plt.legend()
plt.grid(True)

plt.show()
```

```python
products = ['Shampoo', 'Conditioner', 'Hair Oil', 'Bodywash']
sales = [500, 700, 300, 900]

plt.bar(products, sales, color=['red', 'blue', 'green', 'orange'])
plt.xlabel('Products')
plt.ylabel('Sales in Units')
plt.title('Product Sales')
plt.show()
```

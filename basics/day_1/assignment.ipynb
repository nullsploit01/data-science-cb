```python
import pandas as pd
```

```python
csv = pd.read_csv('https://drive.google.com/uc?id=13p-box0F9kou4wE9AyeBNKMSfE767xT-&export=download')
```

```python
country_filter = csv[csv['Country'].isin(['China', 'Russia', 'US'])]
country_filter.to_csv('country_filter_data.csv', index=False)
```

```python
employee_filter = country_filter[
    (country_filter['Number of employees'] > 250) & 
    (country_filter['Number of employees'] < 1000)
]
employee_filter.to_csv('employee_filtered_data.csv', index=False)
```

```python
description_filter = employee_filter[
    employee_filter['Description'].str.contains('food|clothing', case=False, na=False)
]
description_filter.to_csv("description_filter.csv", index=False)
```

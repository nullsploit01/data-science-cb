```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("data.csv")
df = df.dropna()
```

```python
alcohol = df["Alcohol"].values # x
life_expectancy = df["Life expectancy"].values # y

m, b = np.polyfit(alcohol, life_expectancy, 1)
```

```python
def predict(alcohol_value):
    return (m * (alcohol_value + b))
```

```python
predicted_life_expectancy = predict(8.41)
```

```python
print(predicted_life_expectancy)
```

```python
plt.scatter(alcohol, life_expectancy, color="blue", label="Data")
plt.plot(alcohol, predict(alcohol), color="red", label = "Regression Line")
plt.xlabel("Alcohol")
plt.ylabel("Life Expectancy")
plt.legend()
plt.show()
```
